# Project Setup & Data Loading

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"netflix_content_intelligence_combined.csv")
```

# Data Cleaning & Feature Selection

```python
df = df.drop_duplicates()

df = df.dropna(subset = 'date_added')

df['date_added'] = pd.to_datetime(df['date_added'], format='%d-%b-%y', errors='coerce')

fill_values = {
    'rating': 'Unrated',
    'countries': 'Unknown',
    'primary_country': 'Unknown',
    'directors': 'Unknown',
    'cast': 'Unknown'
}
df = df.fillna(value=fill_values)
```

```python
missing = df.isnull().sum()
missing[missing > 0]
```

# Exploratory Data Analysis (EDA)

```python
# How is the distribution of movies versus TV shows,and which ones make up a larger share of Netflix?

type_count = df['type'].value_counts()
type_percent = df['type'].value_counts(normalize=True) * 100

pd.DataFrame({'Count': type_count, 'Percentage %': type_percent})
```

Netflix dedicates approximately 70% of its archive to feature films and about 30% to series.

```python
# Which countries have the most content production on Netflix?

top_countries = df[df['primary_country'] != 'Unknown']['primary_country'].value_counts().head(5)
top_countries_percent = (top_countries / len(df)) * 100

pd.DataFrame({'Count': top_countries , 'Percentage': top_countries_percent})
```

The United States is the most powerful content provider on the platform, accounting for more than 37 percent of the total Netflix archive. India is in second place with 12.28 percent, reflecting Netflix's special focus on the Bollywood film market. The United Kingdom, Canada, and Japan are next in line.

```python
# Users give better ratings to movies or TV shows?

df.groupby('type')['imdb_rating'].agg(['mean' , 'median' , 'count']).round(2)
```

Although movies make up about 70 percent of Netflix's total library, series have been more popular with viewers. The average rating for series (6.94) is significantly higher than for movies (6.21).

```python
# Top 10 Genres Based on Content Quantity and Quality

df[df['primary_genre'] != 'Unknown'].groupby('primary_genre').agg(count = ('type', 'count'), mean_rating = ('imdb_rating' , 'mean')).sort_values(['count' , 'mean_rating'] , ascending = False).head(10).round(2)
```

Although drama and comedy genres make up the majority of Netflix's library, the highest levels of user satisfaction are for documentaries (with an average of 7.03) and crime series (with an average of 7.08). In contrast, horror films, despite their significant production volume, have the lowest average score (5.20).

# Key insights

Content distribution: 69% movies and 31% series.

Quality: Series (average 6.94) have a higher score than movies (6.21).

Countries: America and India have the largest share of production.

Genres: The highest volume is related to Drama and Comedy, but the highest satisfaction is related to Documentary and Crime.
